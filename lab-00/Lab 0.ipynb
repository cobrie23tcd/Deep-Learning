### 💡 Concept note: mutability (why `add_one` and `add_one_to_first` behave differently)

- `add_one(a)`: inside the function, `x = x + 1` **re-binds** the local name `x` to a new integer. The caller's `a` is untouched, so `print(a)` gives `2`.
- `add_one_to_first(l)`: `x[0] = ...` **modifies the object** that both `x` and `l` refer to, so the caller sees `[3, 'hugh']`.
- `add_one(l)` fails because you can't add `1` to a list. `add_one_to_first(a)` fails because an `int` can't be indexed.

**Why this matters in deep learning.** NumPy arrays and PyTorch tensors are mutable. In-place operations such as `X[:,1] = balance/1000` (Lab 2) or `optimizer.step()` (Labs 4–5) change the object for everyone holding a reference to it. That is the point of `optimizer.step()`, but it can also create hard-to-find bugs. Use `.copy()` / `.clone()` when you need an independent copy.

### 💡 NumPy primer (needed from Lab 1 onwards)

Handout-01 writes models in matrix form, $\hat{\mathbf y} = X\mathbf w$. NumPy is how you write that in Python. The most important ideas are:

| Idea | Code | Notes |
|---|---|---|
| Create arrays | `np.array([[1,2],[3,4]])`, `np.zeros((n,p))`, `np.ones(n)`, `np.arange(0,1,0.1)` | |
| Shape | `A.shape` → `(rows, cols)` | Print shapes when debugging. Most ML bugs are shape bugs. |
| Elementwise ops | `A*B`, `A**2`, `np.exp(A)`, `np.log(A)` | |
| Matrix product | `A @ B` or `np.matmul(A,B)` / `np.dot(A,B)` | Inner dimensions must match: `(n,p) @ (p,1) → (n,1)` |
| Transpose | `A.T` | |
| Column / row slicing | `X[:,0]` (column 0), `X[0,:]` (row 0) | |
| Reductions | `np.mean(A)`, `A.sum(axis=0)`, `A.argmax(1)` | |
| Broadcasting | `X - X.mean(axis=0)` | Smaller arrays are stretched to match |
| Linear algebra | `np.linalg.inv(M)`, `np.linalg.solve(M, b)` | Lab 1 normal equation |

Watch out for `(n,)` vs `(n,1)` shapes. Subtracting a `(n,)` array from a `(n,1)` array **broadcasts** to `(n,n)`, which silently gives the wrong result. Use `.reshape(-1,1)` or `.ravel()` to make the shapes match.

The optional cell below lets you try these operations.

````python
# Optional helper added for learning: a quick NumPy warm-up (safe to run, not graded)
import numpy as np

x = np.array([2., 4., 3.])                      # shape (3,)
X = np.stack([x**0, x**1, x**2], axis=1)        # design matrix [1, x, x^2], shape (3, 3)
w = np.array([[1.], [0.5], [-0.1]])             # column vector of weights, shape (3, 1)
y_hat = X @ w                                   # matrix-vector product, shape (3, 1)
print("X =\n", X, "\nshape", X.shape)
print("y_hat =", y_hat.ravel(), "shape", y_hat.shape)

# Broadcasting pitfall: (3,) minus (3,1) gives a (3,3) matrix, not 3 residuals!
print("wrong shape:", (x - y_hat).shape, "  correct shape:", (x.reshape(-1, 1) - y_hat).shape)
print("MSE (vectorised):", np.mean((x.reshape(-1, 1) - y_hat) ** 2))
````

### Exercise 2 explanation

**Objective.** Complete `find_root(f, df, x0)` in `ex_2.py` to approximate a root $\hat{x}$ satisfying $f(\hat{x}) = 0$ using Newton-Raphson optimisation.

Starting from $x_n$, the function evaluates the derivative $f'(x_n)$ and updates the estimate with $x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$. The implementation abandons the search when the derivative is zero, returns when $|x_{n+1}-x_n| < 0.0001$, and stops after more than 100 iterations. These safeguards prevent division by zero and unbounded iteration when the local approximation cannot converge.

Root finding is a core numerical optimisation technique: it appears in model fitting, loss minimisation, and parameter estimation. Its convergence depends on the starting point and local function shape, so the stopping criteria are part of evaluating whether a result is trustworthy. The `%run ex_2.py` cell added below runs the module's `__main__` block, which exercises `find_root` with three test functions, including a case with no real root.

### 🛠️ How to do Exercise 2, step by step (Newton–Raphson)

**Intuition.** Near the current guess $x_n$, replace the curve $f$ by its tangent line $f(x_n) + f'(x_n)(x - x_n)$. Then jump to where that tangent crosses zero:
$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$$
Close to a simple root, the number of correct digits roughly **doubles** each iteration (quadratic convergence).

**Algorithm to implement inside `find_root(f, df, x0)`:**
```python
def find_root(f, df, x0):
    i = 0
    x = x0
    while True:
        derivative = df(x)
        if derivative == 0:                       # 1) tangent is flat: no crossing point
            print("Zero derivative encountered - abandoning search!")
            return None
        next_x = x - f(x) / derivative            # 2) Newton update
        if abs(x - next_x) < 0.0001:              # 3) converged: the step is tiny
            return next_x
        x = next_x
        i += 1
        if i > 100:                               # 4) safety stop: did not converge
            print("Hit max_iterations - abandoning search!")
            return None
```

**Concept link: Newton vs gradient descent.**
- To *minimise* a loss $L(w)$, we look for a root of its gradient, $\nabla L(w) = 0$. Newton's method for that problem uses second derivatives: $w \leftarrow w - [\nabla^2 L]^{-1}\nabla L$.
- Deep learning instead uses the cheaper **gradient descent** $w \leftarrow w - \eta\nabla L$ (handout-02 slides 26–28; Lab 2), because computing and inverting the Hessian is infeasible for millions of parameters.
- The ideas you practise here appear again in Labs 2, 4 and 5: an iterative update, a step size, a convergence test and a maximum number of iterations (epochs).
